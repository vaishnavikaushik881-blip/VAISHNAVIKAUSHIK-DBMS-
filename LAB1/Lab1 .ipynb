```python
def insert_student(sapid, name, mobile, dept, email):
    cur = cnx.cursor()

    query = """
    INSERT INTO xyz1
    (sapid, Nam, Mobile, dept, emailid)
    VALUES (%s, %s, %s, %s, %s)
    """

    values = (sapid, name, mobile, dept, email)

    cur.execute(query, values)
    cnx.commit()   # Save changes

    print("Record inserted successfully.")

    cur.close()
```
